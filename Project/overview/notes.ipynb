# Project Overview – JSON

## 1. What is JSON?

- **JSON (JavaScript Object Notation):** A lightweight data format used to store and exchange data between different applications.
- JSON stores data in **Key-Value pairs**.
- It is commonly used in APIs to transfer data between Frontend and Backend.

## 2. JSON Example

```json
{
    "P001": {
        "id": "P001",
        "name": "Ananya Sharma",
        "city": "Guwahati",
        "age": 28,
        "gender": "female",
        "height": 1.65,
        "weight": 90.0,
        "bmi": 33.06,
        "verdict": "Obese"
    },

    "P002": {
        "id": "P002",
        "name": "Ravi Mehta",
        "city": "Mumbai",
        "age": 35,
        "gender": "male",
        "height": 1.75,
        "weight": 85,
        "bmi": 27.76,
        "verdict": "Overweight"
    }
}
```

## 3. Important JSON Data Types

| Data Type | Example |
|---|---|
| String | `"name": "Ajay"` |
| Integer | `"age": 22` |
| Float | `"height": 1.75` |
| Boolean | `"active": true` |
| Array | `"skills": ["Python", "ML"]` |
| Object | `"user": {"name": "Ajay"}` |
| Null | `"address": null` |

## 5. Frontend Form Fields

- Patient ID
- Name
- City
- Age
- Gender
- Height
- Weight

**SAVE Button:** Sends patient information to the backend for processing and storage.

## 6. AI/ML Perspective

- Patient data can be collected through a frontend form.
- JSON transfers patient information to the backend.
- Backend processes the input data.
- ML model can calculate BMI or predict a result.
- Prediction can be returned to the frontend in JSON format.

**Remember:** JSON is a common data exchange format used by APIs to transfer structured information between Frontend and Backend.
