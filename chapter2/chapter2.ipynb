# FastAPI

- **FastAPI:** A modern, high-performance Python web framework used to build APIs.
- It allows different applications (Web, Android, iOS) to communicate with Python backend code.
- It is widely used to deploy AI/ML models through APIs.

## 1. How FastAPI Works?

```text
Client / Frontend
       |
       | HTTP Request
       v
   Web Server
       |
       v
      ASGI
       |
       v
   FastAPI Code
       |
       v
     ML Model
       |
       v
  Prediction Result
       |
       v
  HTTP Response
       |
       v
Client / Frontend
```

## 2. Main Components

| Component | Work |
|---|---|
| Web Server | Receives HTTP requests |
| ASGI | Connects server with FastAPI |
| FastAPI | Handles API requests |
| ML Model | Generates predictions |
| Response | Returns results to the client |

## 3. What is ASGI?

- **ASGI (Asynchronous Server Gateway Interface):** A standard interface that allows asynchronous Python applications like FastAPI to communicate with an ASGI server.
- Examples: Uvicorn, Hypercorn.

## 4. AI/ML Example

Suppose we have a House Price Prediction Model.

**Request:**
```http
POST /predict
```

**Input (JSON):**
```json
{
    "feature1": 5.2,
    "feature2": 3.1
}
```

**Output (JSON):**
```json
{
    "prediction": 8.3
}
```

## 5. Why is FastAPI Fast?

1. **ASGI Support:** Supports asynchronous request handling.
2. **Async/Await:** Handles waiting operations efficiently.
3. **High Performance:** Built on Starlette and Pydantic.
4. **Data Validation:** Automatically validates input data using Pydantic.
5. **Concurrency:** Handles multiple concurrent requests efficiently.

## 6. Important Terms

| Term | Meaning |
|---|---|
| HTTP | Communication protocol |
| GET | Retrieves data |
| POST | Sends data |
| JSON | Data exchange format |
| ASGI | Interface between server and application |
| Uvicorn | ASGI server |
| Endpoint | Specific API URL |
| Async | Non-blocking asynchronous operations |

**Remember:** FastAPI acts as a bridge between Frontend and ML Model. It receives input, sends it to the model and returns predictions to the client.
