# What is API?

- **API (Application Programming Interface):** Allows two different software applications to communicate and exchange data with each other.

## How API Works?
1. **Client:** Sends a request.
2. **API:** Transfers the request to the server.
3. **Server:** Processes the request.
4. **API:** Receives the server's response.
5. **Client:** Displays the response.

## Real-Life Examples
- **Weather App:** Fetches weather information.
- **Payment App:** Processes online payments.
- **E-commerce:** Manages products, users and orders.
- **AI Applications:** Communicate with AI/ML models.

## Important Terms

| Term | Meaning |
|---|---|
| API | Connects different applications |
| Client | Sends requests |
| Server | Processes requests |
| Request | Instruction sent to server |
| Response | Result returned by server |
| JSON | Data exchange format |

## Example

**Request:**
```http
GET /users/1
```

**Response:**
```json
{
    "id": 1,
    "name": "Ajay",
    "role": "AI Engineer"
}
```

**Remember:** API works like a waiter in a restaurant. It takes requests from the client and brings responses from the server.

## Monolithic Architecture

1. **Monolithic Architecture:** An architecture in which the entire application (Frontend, Backend, Business Logic and Database operations) is developed and deployed as a single unit.
2. All components are tightly connected and work together in one application.
3. **Advantages:** Easy to develop, test and deploy for small applications.
4. **Disadvantages:** As the application grows, it becomes difficult to maintain, scale and update.
5. **Why less popular for large applications?** Modern applications need independent scaling, faster updates and flexibility, so Microservices Architecture is often preferred over Monolithic Architecture.

**Note:** Monolithic Architecture is still used today, especially for small and medium-sized applications. It is not outdated.

## Monolithic Architecture – Folder Structure

```text
my_application/
│
├── frontend/
│   ├── web/
│   ├── android/
│   └── ios/
│
├── backend/
│   ├── routes/
│   ├── controllers/
│   ├── models/
│   ├── services/
│   ├── middleware/
│   ├── database.py
│   └── main.py
│
├── ml_model/
│   ├── trained_model.pkl
│   └── predict.py
│
├── requirements.txt
├── .env
└── README.md
```

**Explanation:**
- `frontend/` – Handles user interfaces.
- `backend/` – Handles API requests and business logic.
- `ml_model/` – Handles ML model predictions.
- `database.py` – Manages database connection.
- `main.py` – Entry point of the backend application.

**Remember:** In Monolithic Architecture, all application components are maintained and deployed as one application.

## ML API – Monolithic Architecture

```text
my_ml_api/
│
├── src/
│   ├── templates/
│   │   └── index.html
│   │
│   ├── static/
│   │   ├── style.css
│   │   └── script.js
│   │
│   ├── model.pkl
│   ├── ml_backend.py
│   └── app.py
```

### Folder & File Explanation

| Component | Purpose |
|---|---|
| `src/` | Main source code folder |
| `templates/` | Stores HTML frontend files |
| `static/` | Stores CSS and JavaScript files |
| `model.pkl` | Stores the trained ML model |
| `ml_backend.py` | Handles ML model loading and prediction |
| `app.py` | Main application and API handling |
